```python
im = pyg.screenshot()
c = im.getpixel((174,320))
print(c)

```

```python
target_color = (83, 83, 83)
was_target_color = False
while True:
    im = pyg.screenshot()
    c = im.getpixel((174, 320))
    is_target_color = c == target_color
    if is_target_color and not was_target_color:
        pyg.press("space")
    was_target_color = is_target_color
```
