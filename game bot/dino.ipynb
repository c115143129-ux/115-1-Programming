```python
import pyautogui as pyg
import time

x = 240
b = 0

# 記錄程式開始時間與上次移動時間
start = time.time()
last = start

mode = "light"

while True:
    im = pyg.screenshot()
    now = time.time()

    if now - start < 20:
        x = 240
    elif now - start > 20 and now - start < 30:
        x = 350
    elif now - start > 30 and now - start < 40:
        x = 390
    elif now - start > 40 and now - start < 60:
        x = 450
    else:
        x = 520     

    # 偵測障礙物
    for i in range(40):
        for j in range(60):
            c = im.getpixel((x + i, 690 + j))
            if c != (255, 255, 255) and c != (0, 0, 0):

                print(c)
            if c == (81,81,81)or c ==(172,172,172)or c ==(83,83,83) or c == (82,82,82):
                
                b += 1
                pyg.press("space")
                pyg.sleep(0.1)
                break
        else:
            continue
        break
```
