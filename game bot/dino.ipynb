```python
import pyautogui as pyg
import time

b = 0
start = time.time()

# 基準 Y 軸與高度（請依據螢幕視窗位置調整）
TOP_Y = 615
HEIGHT = 60

print("小恐龍自動腳本啟動中... (請確保遊戲視窗在前台)")

while True:
    now = time.time()
    elapsed = now - start

    # 1. 時間判斷邏輯修復（改用簡化邊界，避免剛好整數秒時落在 else 的 bug）
    if elapsed < 20:
        x = 280
        scan_width = 40
    elif elapsed < 30:
        x = 350
        scan_width = 50
    elif elapsed < 40:
        x = 450
        scan_width = 60
    elif elapsed < 60:
        x = 460
        scan_width = 70
    else:
        x = 540
        scan_width = 80

    # 2. 區域截圖：只擷取前方障礙物偵測框（大幅提升截圖速度）
    im = pyg.screenshot(region=(x, 615, scan_width, 60))

    # 3. 取得區域左上角 (0, 0) 作為背景基準色（自動支援白天 / 夜間模式）
    bg_color = im.getpixel((0, 0))

    has_obstacle = False

    # 4. 步進抽樣掃描 (step=8)，大幅減少點對點比對次數
    for i in range(0, scan_width, 8):
        for j in range(0, HEIGHT, 8):
            pixel = im.getpixel((i, j))
            
            # 計算與背景色的差異 (只要有 RGB 差異超過 15 視為障礙物)
            if abs(pixel[0] - bg_color[0]) > 15 or abs(pixel[1] - bg_color[1]) > 15 or abs(pixel[2] - bg_color[2]) > 15:
                has_obstacle = True
                break
        if has_obstacle:
            break

    # 5. 偵測到障礙物即跳躍
    if has_obstacle:
        b += 1
        pyg.press("space")
        # 短暫冷卻避免連續誤按
        time.sleep(0.15)
```
